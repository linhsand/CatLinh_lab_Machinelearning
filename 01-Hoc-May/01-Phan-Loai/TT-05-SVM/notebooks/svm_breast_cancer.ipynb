```python
#import thư viện
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import (
    train_test_split,
    GridSearchCV
)

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve
)

import warnings
warnings.filterwarnings("ignore")
```

```python
#load dataset
data = load_breast_cancer(as_frame=True)

X = data.data
y = data.target

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#kiểm tra nhãn
print(data.target_names)
print(y.value_counts())
```

```python
df = X.copy()
df["target"] = y

df.head()
df.shape
df.info()
df.isnull().sum().sum()
```

```python
#kiểm tra phân bổ target
sns.countplot(x=y)

plt.title("Distribution of Target")
plt.xlabel("Target")
plt.ylabel("Count")

plt.show()
```

```python
#vẽ correlation heatmap ( bỏ biến target)
plt.figure(figsize=(16, 12))

corr = X.corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of 30 Features")
plt.show()
```

```python
#chia train/test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)


print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\ny_train:")
print(y_train.value_counts())

print("\ny_test:")
print(y_test.value_counts())
```

```python
#SVM KHÔNG SCALE
svm_no_scale = SVC(
    kernel="rbf",
    C=1,
    gamma="scale",
    random_state=42
)

svm_no_scale.fit(X_train, y_train)

y_pred_no_scale = svm_no_scale.predict(X_test)

print("Accuracy:",
      accuracy_score(y_test, y_pred_no_scale))

print("Recall:",
      recall_score(
          y_test,
          y_pred_no_scale,
          pos_label=0
      ))

print("Precision:",
      precision_score(
          y_test,
          y_pred_no_scale,
          pos_label=0
      ))
```

```python
#SVM SCALE
svm_scaled = Pipeline([
    ("scale", StandardScaler()),
    ("svm", SVC(
        kernel="rbf",
        C=1,
        gamma="scale",
        random_state=42
    ))
])

svm_scaled.fit(X_train, y_train)

y_pred_scaled = svm_scaled.predict(X_test)

print("Accuracy:",
      accuracy_score(y_test, y_pred_scaled))

print("Recall malignant:",
      recall_score(
          y_test,
          y_pred_scaled,
          pos_label=0
      ))

print("Precision malignant:",
      precision_score(
          y_test,
          y_pred_scaled,
          pos_label=0
      ))
```

```python
#tạo bảng so sánh
results_scale = pd.DataFrame({
    "Model": [
        "SVM - No Scaling",
        "SVM - StandardScaler"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred_no_scale),
        accuracy_score(y_test, y_pred_scaled)
    ],
    "Precision_Malignant": [
        precision_score(y_test, y_pred_no_scale, pos_label=0),
        precision_score(y_test, y_pred_scaled, pos_label=0)
    ],
    "Recall_Malignant": [
        recall_score(y_test, y_pred_no_scale, pos_label=0),
        recall_score(y_test, y_pred_scaled, pos_label=0)
    ],
    "F1_Malignant": [
        f1_score(y_test, y_pred_no_scale, pos_label=0),
        f1_score(y_test, y_pred_scaled, pos_label=0)
    ]
})

results_scale
```

```python
#so sánh 3 kernel
kernels = ["linear", "rbf", "poly"]

kernel_results = []

for kernel in kernels:
    
    model = Pipeline([
        ("scale", StandardScaler()),
        ("svm", SVC(
            kernel=kernel,
            C=1,
            random_state=42
        ))
    ])
    
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    
    kernel_results.append({
        "Kernel": kernel,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision_Malignant": precision_score(
            y_test, pred, pos_label=0
        ),
        "Recall_Malignant": recall_score(
            y_test, pred, pos_label=0
        ),
        "F1_Malignant": f1_score(
            y_test, pred, pos_label=0
        )
    })
    
    
kernel_results = pd.DataFrame(kernel_results)

kernel_results
```

```python
#gridsearchCV
from sklearn.metrics import make_scorer

malignant_recall = make_scorer(
    recall_score,
    pos_label=0
)

pipe = Pipeline([
    ("scale", StandardScaler()),
    ("svm", SVC(
        probability=True,
        class_weight="balanced",
        random_state=42
    ))
])

param_grid = {
    "svm__C": [0.1, 1, 10, 100],
    "svm__gamma": ["scale", 0.001, 0.01, 0.1],
    "svm__kernel": ["linear", "rbf", "poly"]
}


gs = GridSearchCV(
    pipe,
    param_grid,
    cv=5,
    scoring=malignant_recall,
    n_jobs=-1
)

gs.fit(X_train, y_train)

print("Best parameters:")
print(gs.best_params_)

print("\nBest CV malignant recall:")
print(gs.best_score_)
```

```python
# Heatmap C x Gamma cho RBF kernel

cv_results = pd.DataFrame(gs.cv_results_)

# Chỉ lấy kết quả của RBF
rbf_results = cv_results[
    cv_results["param_svm__kernel"] == "rbf"
].copy()

# Chuyển gamma thành text để làm index/column
rbf_results["gamma"] = rbf_results["param_svm__gamma"].astype(str)
rbf_results["C"] = rbf_results["param_svm__C"].astype(float)

# Tạo bảng C x gamma
heatmap_data = rbf_results.pivot(
    index="C",
    columns="gamma",
    values="mean_test_score"
)

heatmap_data
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".4f",
    cmap="YlOrRd"
)

plt.title("SVM RBF - C × Gamma\nMean CV Recall for Malignant Class")
plt.xlabel("Gamma")
plt.ylabel("C")

plt.tight_layout()

plt.savefig(
    "../reports/C_gamma_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Best model từ GridSearchCV

best_model = gs.best_estimator_

# Predict trên test set
y_pred_best = best_model.predict(X_test)

print("Best parameters:")
print(gs.best_params_)

print("\nTest Accuracy:")
print(accuracy_score(y_test, y_pred_best))

print("\nTest Precision - Malignant:")
print(
    precision_score(
        y_test,
        y_pred_best,
        pos_label=0
    )
)

print("\nTest Recall - Malignant:")
print(
    recall_score(
        y_test,
        y_pred_best,
        pos_label=0
    )
)

print("\nTest F1 - Malignant:")
print(
    f1_score(
        y_test,
        y_pred_best,
        pos_label=0
    )
)
```

```python
from sklearn.model_selection import cross_val_predict

# Lấy xác suất dự đoán bằng 5-fold CV trên training set
oof_proba = cross_val_predict(
    best_model,
    X_train,
    y_train,
    cv=5,
    method="predict_proba",
    n_jobs=-1
)

# Xác suất thuộc class 0 = malignant
oof_malignant_proba = oof_proba[:, 0]

print("OOF probabilities shape:", oof_malignant_proba.shape)
```

```python
#tìm threshold đạt recall >=0.98
threshold_results = []

for threshold in np.arange(0.05, 0.51, 0.01):
    
    # Nếu xác suất malignant >= threshold
    # => dự đoán malignant = 0
    y_pred_threshold = np.where(
        oof_malignant_proba >= threshold,
        0,
        1
    )
    
    recall = recall_score(
        y_train,
        y_pred_threshold,
        pos_label=0
    )
    
    precision = precision_score(
        y_train,
        y_pred_threshold,
        pos_label=0
    )
    
    threshold_results.append({
        "Threshold": threshold,
        "Recall_Malignant": recall,
        "Precision_Malignant": precision
    })

threshold_df = pd.DataFrame(threshold_results)

threshold_df
```

```python
valid_thresholds = threshold_df[
    threshold_df["Recall_Malignant"] >= 0.98
].copy()

valid_thresholds
```

```python
best_threshold_row = valid_thresholds.iloc[
    valid_thresholds["Threshold"].idxmax()
]

best_threshold = best_threshold_row["Threshold"]

print("Selected threshold:", best_threshold)
print(
    "OOF Recall:",
    best_threshold_row["Recall_Malignant"]
)
print(
    "OOF Precision:",
    best_threshold_row["Precision_Malignant"]
)

#tại sao chọn threshold cao nhất? --> cần recall >0.98 -> chọn threshold đạt mục tiêu recall nhưng precision cao hơn
```

```python
#vẽ recall và precision theo threshold
plt.figure(figsize=(10, 6))

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Recall_Malignant"],
    marker="o",
    label="Recall - Malignant"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Precision_Malignant"],
    marker="o",
    label="Precision - Malignant"
)

plt.axhline(
    0.98,
    linestyle="--",
    label="Recall target = 0.98"
)

plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Selected threshold = {best_threshold:.2f}"
)

plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Precision-Recall Trade-off by Threshold")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
#đánh giá threshold trên test set
test_proba = best_model.predict_proba(X_test)

# Probability của malignant = class 0
test_malignant_proba = test_proba[:, 0]

y_pred_threshold = np.where(
    test_malignant_proba >= best_threshold,
    0,
    1
)

threshold_recall = recall_score(
    y_test,
    y_pred_threshold,
    pos_label=0
)

threshold_precision = precision_score(
    y_test,
    y_pred_threshold,
    pos_label=0
)

threshold_f1 = f1_score(
    y_test,
    y_pred_threshold,
    pos_label=0
)

threshold_accuracy = accuracy_score(
    y_test,
    y_pred_threshold
)

print("Selected threshold:", best_threshold)

print("\nTest Accuracy:", threshold_accuracy)
print("Test Precision - Malignant:", threshold_precision)
print("Test Recall - Malignant:", threshold_recall)
print("Test F1 - Malignant:", threshold_f1)
```

### Support Vectors — Interpretation

The best SVM model has 184 support vectors out of 455 training samples,
representing 40.44% of the training data.

There are 94 support vectors from the malignant class and 90 from the
benign class.

Support vectors are the observations closest to the decision boundary.
They are the most important observations for determining the position
of the SVM decision boundary, while observations farther away have less
influence on the boundary.

Therefore, the model's decision boundary is determined by a substantial
portion of the training data, with relatively similar contributions from
the two classes.
